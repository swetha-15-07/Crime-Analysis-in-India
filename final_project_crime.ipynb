**Stage 3 – EDA and Visualizations**

The goal is to understand patterns, identify unusual values, compare variables, and explain what the results mean for a business.

```python

import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel("project 11.xlsx")

# Identify crime columns, excluding year and ID codes
crime_cols = [
    c for c in df.select_dtypes(include="number").columns
    if c not in [
        "Year (year)", "State Code (state_code)",
        "District Code (district_code)"
    ] and df[c].notna().any()
]

# Remove columns that are entirely missing
df = df.dropna(axis=1, how="all")
crime_cols = [c for c in crime_cols if c in df.columns]

# Calculate total recorded crimes per row
df["Total_Crimes"] = df[crime_cols].sum(axis=1)

sns.set_theme(style="whitegrid")



```

```python
# 1. Histogram: distribution of total crimes
plt.figure(figsize=(8, 5))
sns.histplot(df["Total_Crimes"], bins=30, kde=True)
plt.title("Distribution of Total Recorded Crimes")
plt.xlabel("Total recorded crimes per record")
plt.ylabel("Number of records")
plt.tight_layout()
plt.show()

mean_val = df["Total_Crimes"].mean()
median_val = df["Total_Crimes"].median()
print("INTERPRETATION:")
print(f"Mean total = {mean_val:.2f}; median = {median_val:.2f}.")
if mean_val > median_val:
    print("The distribution may be right-skewed: a smaller number of "
          "records have comparatively high totals.")
elif mean_val < median_val:
    print("The distribution may be left-skewed.")
else:
    print("Mean and median are similar; inspect the chart for the shape.")
```

```python
# 2. Boxplot: identify unusually high totals
plt.figure(figsize=(8, 3))
sns.boxplot(x=df["Total_Crimes"])
plt.title("Boxplot of Total Recorded Crimes")
plt.xlabel("Total recorded crimes per record")
plt.tight_layout()
plt.show()

q1 = df["Total_Crimes"].quantile(0.25)
q3 = df["Total_Crimes"].quantile(0.75)
iqr = q3 - q1
outliers = df[
    (df["Total_Crimes"] < q1 - 1.5 * iqr) |
    (df["Total_Crimes"] > q3 + 1.5 * iqr)
]
print("INTERPRETATION:")
print(f"{len(outliers)} records fall outside the 1.5×IQR limits.")
print("These records need investigation; unusually high values are not "
      "automatically data errors.")
```

```python


# 3. Barplot: most frequently recorded crime categories
crime_totals = df[crime_cols].sum().sort_values(ascending=False)
top_crimes = crime_totals.head(10)

plt.figure(figsize=(11, 6))
sns.barplot(x=top_crimes.values, y=top_crimes.index)
plt.title("Top 10 Crime Categories by Recorded Count")
plt.xlabel("Total recorded count")
plt.ylabel("Crime category")
plt.tight_layout()
plt.show()

print("INTERPRETATION:")
print("Highest-count category:", top_crimes.index[0])
print("Recorded count:", int(top_crimes.iloc[0]))
print("Business meaning: this identifies categories contributing the "
      "largest recorded counts and helps prioritize further investigation.")
```

```python

# 4. Scatterplot: riots versus other IPC crimes
riots_col = next(
    (c for c in crime_cols if "riots" in c.lower()), None
)
other_col = next(
    (c for c in crime_cols if "other_ipc_crimes" in c.lower()), None
)

if riots_col and other_col:
    plt.figure(figsize=(8, 5))
    sns.scatterplot(
        data=df, x=riots_col, y=other_col,
        alpha=0.7
    )
    plt.title("Riots vs Other IPC Crimes")
    plt.xlabel("Recorded riots")
    plt.ylabel("Other IPC crimes")
    plt.tight_layout()
    plt.show()

    corr = df[[riots_col, other_col]].corr().iloc[0, 1]
    print("INTERPRETATION:")
    print(f"Correlation = {corr:.2f}")
    if corr > 0:
        print("The variables have a positive linear association.")
    elif corr < 0:
        print("The variables have a negative linear association.")
    else:
        print("There is no linear association in this sample.")
    print("Association does not establish causation.")

```

```python
#5. Correlation heatmap
selected_cols = crime_cols[:]
corr_matrix = df[selected_cols].corr()

plt.figure(figsize=(12, 9))
sns.heatmap(
    corr_matrix, cmap="coolwarm", center=0,
    vmin=-1, vmax=1
)
plt.title("Correlation Between Crime Categories")
plt.tight_layout()
plt.show()

print("INTERPRETATION:")
pairs = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
).stack()
pairs = pairs.dropna()
pairs = pairs[pairs.index.get_level_values(0) !=
              pairs.index.get_level_values(1)]

if not pairs.empty:
    strongest_pair = pairs.abs().idxmax()
    print("Strongest absolute correlation:",
          strongest_pair[0], "and", strongest_pair[1])
    print(f"Correlation = {pairs.loc[strongest_pair]:.2f}")
    print("Business meaning: strongly associated categories may "
          "deserve joint investigation, but correlation is not causation.")

```

```python
# 6. Compare recorded crime totals across states/regions
state_col = "State Name (state_name)"
if state_col in df.columns:
    state_totals = (
        df.groupby(state_col)["Total_Crimes"]
        .sum().sort_values(ascending=False).head(10)
    )

    plt.figure(figsize=(10, 6))
    sns.barplot(x=state_totals.values, y=state_totals.index)
    plt.title("Top 10 States/Regions by Recorded Crime Total")
    plt.xlabel("Total recorded count")
    plt.ylabel("State/region")
    plt.tight_layout()
    plt.show()

    print("INTERPRETATION:")
    print("Highest total:", state_totals.index[0],
          "-", int(state_totals.iloc[0]))
    print("This compares recorded totals, not population-adjusted "
          "crime rates. Differences may also reflect coverage and reporting.")
```

```python
# 7. Pivot table: state/region versus crime category
state_col = "State Name (state_name)"

if state_col in df.columns:
    pivot = pd.pivot_table(
        df,
        index=state_col,
        values=crime_cols,
        aggfunc="sum",
        fill_value=0
    )

    print("State-wise crime category pivot table:")
    print(pivot.head(10))

    # Select the five largest crime categories overall
    top5_cols = crime_totals.head(5).index.tolist()

    # Show the top 10 states/regions for these categories
    top_states = (
        df.groupby(state_col)["Total_Crimes"]
        .sum().nlargest(10).index
    )

    heatmap_data = pivot.loc[top_states, top5_cols]

    plt.figure(figsize=(12, 7))
    sns.heatmap(heatmap_data, annot=True, fmt=".0f",
                cmap="YlOrRd")
    plt.title("Crime Category Comparison Across Top 10 States/Regions")
    plt.xlabel("Crime category")
    plt.ylabel("State/region")
    plt.tight_layout()
    plt.show()

    print("INTERPRETATION:")
    print("The heatmap shows which crime categories contribute most "
          "to recorded counts in each selected state/region.")
    print("Business use: identify where category-specific review, "
          "data validation, or additional investigation may be useful.")
```

```python
# 8. Grouped analysis: district-level totals within states
district_col = "District Name (district_name)"

if state_col in df.columns and district_col in df.columns:
    grouped = (
        df.groupby([state_col, district_col], as_index=False)
        ["Total_Crimes"].sum()
    )

    print("Highest recorded district totals:")
    print(grouped.nlargest(10, "Total_Crimes").to_string(index=False))

    print("INTERPRETATION:")
    print("The table highlights district records with the highest "
          "recorded totals. Validate the records and examine crime "
          "categories before drawing conclusions.")
```

```python
# 9. Pairplot: relationships among selected crime variables
pair_cols = crime_totals.head(4).index.tolist()

if len(pair_cols) >= 2:
    pair_df = df[pair_cols].copy()

    sns.pairplot(pair_df, corner=True, diag_kind="hist")
    plt.suptitle(
        "Pairplot of the Four Highest-Count Crime Categories",
        y=1.02
    )
    plt.show()

    print("INTERPRETATION:")
    print("The diagonal plots show each category's distribution.")
    print("The off-diagonal plots show whether categories move "
          "together and whether unusual records are present.")
    print("Use this to identify patterns for further investigation, "
          "not to infer cause and effect.")
```
